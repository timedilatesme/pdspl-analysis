```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# FIGURE 1: Power-Law Density Profile
# ==========================================

# Set dark background to match your presentation
plt.style.use('dark_background')

# Radius array
r = np.linspace(0.1, 5, 500)

# Density profiles for different gamma_pl
gamma_isothermal = 2.0
gamma_steep = 2.3
gamma_shallow = 1.7

rho_iso = r**(-gamma_isothermal)
rho_steep = r**(-gamma_steep)
rho_shallow = r**(-gamma_shallow)

# Create figure
fig1, ax1 = plt.subplots(figsize=(5, 4), dpi=300)

# Plot lines
ax1.plot(r, rho_shallow, color='#4DAF4A', lw=3, label=r'$\gamma_{pl} = 1.7$ (Shallow)')
ax1.plot(r, rho_iso, color='#FFFF33', lw=3, ls='--', label=r'$\gamma_{pl} = 2.0$ (Isothermal)')
ax1.plot(r, rho_steep, color='#E41A1C', lw=3, label=r'$\gamma_{pl} = 2.3$ (Steep)')

# Explicitly add the governing equation to the plot
ax1.text(0.55, 0.85, r'$\rho(r) \propto r^{-\gamma_{pl}}$', 
         transform=ax1.transAxes, fontsize=16, fontweight='bold', color='white',
         verticalalignment='top', bbox=dict(boxstyle='round,pad=0.4', facecolor='#222222', alpha=0.8, edgecolor='gray'))

# Formatting
ax1.set_xscale('log')
ax1.set_yscale('log')
ax1.set_xlabel('Radius $r$', fontsize=14, fontweight='bold')
ax1.set_ylabel(r'Density $\rho(r)$', fontsize=14, fontweight='bold')
ax1.set_title('Power-Law Density Profile', fontsize=16, fontweight='bold', pad=15)
ax1.legend(fontsize=11, frameon=False, loc='lower left')

# Remove borders for a cleaner presentation look
ax1.spines['top'].set_visible(False)
ax1.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()


# ==========================================
# FIGURE 2: Mass-Sheet Degeneracy (MST)
# ==========================================

import numpy as np
import matplotlib.pyplot as plt

plt.style.use('dark_background')

# Radius array
r = np.linspace(0.1, 4, 500)

# True convergence (surface mass density) - assuming an isothermal profile (1/r)
kappa_true = 1.0 / r

# MST parameters
lambda_mst = 0.6  # Scaling factor
mass_sheet = 1.0 - lambda_mst  # The uniform sheet (0.4)

# Transformed convergence: scaled galaxy + flat sheet
kappa_transformed = (lambda_mst * kappa_true) + mass_sheet

# Create figure
fig2, ax2 = plt.subplots(figsize=(5, 4), dpi=300)

# Plot the True Galaxy
ax2.plot(r, kappa_true, color='#FFFFFF', lw=3, ls='--', 
        label=r'Original Lens ($\lambda = 1.0$)')

# Plot the Transformed Galaxy
ax2.plot(r, kappa_transformed, color='#FF9900', lw=3, 
        label=r'MST Lens ($\lambda = 0.6$)')

# Fill the uniform mass sheet area to make it visually obvious
ax2.fill_between(r, 0, mass_sheet, color='#00BFFF', alpha=0.3)
ax2.axhline(mass_sheet, color='#00BFFF', lw=2, ls=':', 
           label=r'Uniform Mass Sheet ($1 - \lambda$)')

# Fill the scaled galaxy sitting on top of the sheet
ax2.fill_between(r, mass_sheet, kappa_transformed, color='#FF9900', alpha=0.15)

# Explicitly add the governing equation to the plot
ax2.text(0.5, 0.88, r'$\kappa_{trans}(r) = \lambda \kappa_{true}(r) + (1 - \lambda)$', 
         transform=ax2.transAxes, fontsize=13, fontweight='bold', color='white',
         horizontalalignment='center', bbox=dict(boxstyle='round,pad=0.4', facecolor='#222222', alpha=0.8, edgecolor='gray'))


# Formatting
ax2.set_xlim(0.1, 4)
ax2.set_ylim(0, 3.5)
ax2.set_xlabel('Radius $r$', fontsize=14, fontweight='bold')
ax2.set_ylabel(r'Surface Mass Density $\kappa(r)$', fontsize=14, fontweight='bold')
ax2.set_title('The Mass-Sheet Degeneracy', fontsize=16, fontweight='bold', pad=15)

# Legend placement
ax2.legend(fontsize=10, frameon=False, loc='center right')

# Remove borders for a cleaner presentation look
ax2.spines['top'].set_visible(False)
ax2.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```
