```python
import pickle
import numpy as np
import matplotlib.pyplot as plt
import corner
import matplotlib.lines as mlines
import os

# Import plotting utility
from pdspl_utils.plotting import plot_dspl_corner

# Directories for saving figures
FIGURE_DIRECTORY = '../figures/forecasts/forecast_paper_figures'
os.makedirs(FIGURE_DIRECTORY, exist_ok=True)

# Load the saved posterior results
with open('../data/samples/forecast_scenarios_w0waCDM_fixed_scatter.pkl', 'rb') as f:
    fixed_scenarios = pickle.load(f)

with open('../data/samples/forecast_scenarios_w0waCDM_free_scatter_lsst_y10.pkl', 'rb') as f:
    free_scenarios = pickle.load(f)
```

```python
# --- TRUTH VALUES & FIXED PARAMS ---
# We define the core 7 sampled cosmological and population parameters
truth = {
    'h0': 70.0, 'om': 0.3, 'w0': -1.0, 'wa': 0.0,
    'lambda_int': 1.0, 'lambda_sigma': 0.05,
    'gamma_pl': 2.0, 'gamma_sigma': 0.16
}

fixed_params = {'h0': 70.0}

latex_labels = {
    'om': r"$\Omega_{\rm m}$", 'w0': r"$w_0$", 'wa': r"$w_a$", 
    'lambda_int': r"$\overline{\lambda}_{\rm MST}$", 
    'lambda_sigma': r"$\sigma({\lambda}_{\rm MST})$",
    'gamma_pl': r"$\overline{\gamma}_{\rm pl}$", 
    'gamma_sigma': r"$\sigma({\gamma}_{\rm pl})$",
    'beta_c0': r"$\sigma_{\beta_{\rm E},\rm \mathcal{D}}^{(0)}$",
    'beta_c1': r"$\sigma_{\beta_{\rm E},\rm \mathcal{D}}^{(1)}$",
    'beta_c2': r"$\sigma_{\beta_{\rm E},\rm \mathcal{D}}^{(2)}$"
}

custom_ranges = [
    (0, 1),            # om
    (-2, 0),       # w0
    (-3, 3),           # wa
    (truth['lambda_int'] - 0.2, truth['lambda_int'] + 0.2), # lambda_int
    (0.0, 0.18),       # lambda_sigma
    (truth['gamma_pl'] - 0.2, truth['gamma_pl'] + 0.2),     # gamma_pl
    (0.0, 0.32),      # gamma_sigma
]
```

## 1. Comparison of all PDSPL Fixed Scatter Forecasts
Excluding joint datasets, visualizing the impact of spectroscopic vs photometric redshifts across various PDSPL subsets.

```python
scenarios_to_plot_fig1 = [
    'lsst_4most_spec-z_sigma_v', 
    'lsst_4most_spec-z', 
    'lsst_y1', 
    'lsst_y10'
]


custom_linestyles_fig1 = {
    'lsst_4most_spec-z_sigma_v': '--',
    'lsst_4most_spec-z': '--',
    'lsst_y1': '-',
    'lsst_y10': '-'
}

fig1 = plot_dspl_corner(
    scenarios=fixed_scenarios, 
    truth=truth, 
    fixed_params=fixed_params, 
    scenarios_to_plot=scenarios_to_plot_fig1,
    custom_ranges=custom_ranges,
    latex_labels=latex_labels,
    figsize=(14, 14),
    show_multiple_titles=True,
    custom_linestyles_dict=custom_linestyles_fig1,
    save_path=f"{FIGURE_DIRECTORY}/all_pdspls_forecast_w0waCDM_fixed_scatter.pdf"
)
plt.show()
```

## 2. PDSPL LSST Y10 vs DSPL (500 lenses)
Comparing realistic PDSPL selections against an optimistic, synthetic DSPL population, both with and without $\Omega_{\rm m}$ priors.

```python
scenarios_to_plot_fig2 = [
    'DSPL',
    'lsst_y10',
    'DSPL_om_prior', 
    'lsst_y10_om_prior'
]

custom_colors_fig2 = {
    'DSPL': fixed_scenarios['DSPL']['color'],
    'lsst_y10': fixed_scenarios['lsst_y10']['color'],
    'DSPL_om_prior': '#002B5F',  # A Darker Blue
    'lsst_y10_om_prior': '#135413'  # Dark green
}

custom_linestyles_fig2 = {
    'DSPL': '--',
    'lsst_y10': '-',
    'DSPL_om_prior': '--',
    'lsst_y10_om_prior': '-'
}

fig2 = plot_dspl_corner(
    scenarios=fixed_scenarios, 
    truth=truth, 
    fixed_params=fixed_params, 
    scenarios_to_plot=scenarios_to_plot_fig2,
    custom_ranges=custom_ranges,
    latex_labels=latex_labels,
    figsize=(14, 14),
    show_multiple_titles=True,
    custom_colors_dict=custom_colors_fig2,
    custom_linestyles_dict=custom_linestyles_fig2,
    save_path=f"{FIGURE_DIRECTORY}/pdspl_vs_dspl_forecast_w0waCDM_fixed_scatter.pdf"
)
plt.show()
```

## 3. Dark Energy Focus: $w_0$-$w_a$ Contours
A zoomed-in, 2-parameter contour plot focusing exclusively on the Dark Energy equation of state parameters to showcase the constraining power of Joint scenarios.

```python
# Create a specialized scenario dictionary strictly for w0 and wa
w0wa_scenarios = {}
keys_fig3 = ['lsst_y10', 'DSPL']#, 'joint_PDSPL_DSPL']

for key in keys_fig3:
    sc = fixed_scenarios[key]
    w0wa_scenarios[key] = {
        'name': sc['name'],
        'color': sc['color'],
        # Slice indices 1 and 2, which correspond to w0 and wa
        'samples': sc['samples'][:, 1:3] 
    }

# Define isolated truths and ranges
truth_w0wa = {'w0': truth['w0'], 'wa': truth['wa']}
custom_ranges_w0wa = [custom_ranges[1], custom_ranges[2]]

fig3 = plot_dspl_corner(
    scenarios=w0wa_scenarios, 
    truth=truth_w0wa, 
    fixed_params={}, # Empty because we sliced out the others
    custom_ranges=custom_ranges_w0wa,
    latex_labels=latex_labels,
    figsize=(8, 8),
    show_multiple_titles=True,
    save_path=f"{FIGURE_DIRECTORY}/w0_wa_joint_focus.pdf"
)
plt.show()
```

## 4. Free vs. Fixed Scatter Comparison (LSST Y10)
Merging the data from `04a` and `04b` to see how allowing the empirical scatter coefficients to vary impacts our cosmological constraints.

```python
merged_scenarios = {}

# 1. Add Fixed Scenarios (Slice first 7 columns to be safe)
for key in ['lsst_y10', 'lsst_y10_om_prior']:
    sc = fixed_scenarios[key]
    merged_scenarios[f"fixed_{key}"] = {
        'name': sc['name'] + " (Fixed Scat.)",
        'color': '#2ca02c' if 'om' not in key else '#135413',
        'samples': sc['samples'][:, :7]
    }

# 2. Add Free Scenarios
# The free samples have 9 columns (om, w0, wa, lambda, l_sig, gamma, g_sig, beta_c0, beta_c1).
# We slice the first 7 to project onto the shared parameter space for direct comparison.
for key in ['lsst_y10', 'lsst_y10_om_prior']:
    sc = free_scenarios[key]
    merged_scenarios[f"free_{key}"] = {
        'name': sc['name'] + " (Free Scat.)",
        'color': '#ff7f0e' if 'om' not in key else '#a63603', # Orange / Dark Orange
        'samples': sc['samples'][:, :7]
    }

scenarios_to_plot_fig4 = [
    'fixed_lsst_y10', 
    'free_lsst_y10', 
    'fixed_lsst_y10_om_prior', 
    'free_lsst_y10_om_prior'
]

custom_linestyles_fig4 = {
    'fixed_lsst_y10': '-',
    'free_lsst_y10': '--',
    'fixed_lsst_y10_om_prior': '-',
    'free_lsst_y10_om_prior': '--'
}

fig4 = plot_dspl_corner(
    scenarios=merged_scenarios, 
    truth=truth, 
    fixed_params=fixed_params, 
    scenarios_to_plot=scenarios_to_plot_fig4,
    custom_ranges=custom_ranges,
    latex_labels=latex_labels,
    figsize=(14, 14),
    show_multiple_titles=True,
    custom_linestyles_dict=custom_linestyles_fig4,
    save_path=f"{FIGURE_DIRECTORY}/free_vs_fixed_scatter_lsst_y10_forecast_w0waCDM.pdf"
)
plt.show()
```

## 5. Free Scatter Forecast (LSST Y10)

```python
scenarios_to_plot_fig5 = [
    'lsst_y10', 
    'lsst_y10_om_prior'
]

custom_linestyles_fig5 = {
    'lsst_y10': '-',
    'lsst_y10_om_prior': '-'
}

custom_ranges_free = custom_ranges + [
    (0, 0.25),     # beta_c0 
    (0, 2.0),       # beta_c1 
]

custom_truth_free = {
    'om': truth['om'], 
    'w0': truth['w0'], 
    'wa': truth['wa'], 
    'lambda_int': truth['lambda_int'], 
    'lambda_sigma': truth['lambda_sigma'], 
    'gamma_pl': truth['gamma_pl'], 
    'gamma_sigma': truth['gamma_sigma'],
    'beta_c0': 0.14960267, # From the pdspl_samples pickle file for LSST Y10
    'beta_c1': 0.89311862
}

custom_colors_fig5 = {
    'lsst_y10': merged_scenarios['free_lsst_y10']['color'],
    'lsst_y10_om_prior': merged_scenarios['free_lsst_y10_om_prior']['color']
}

custom_free_scenarios = free_scenarios.copy() 
custom_free_scenarios['lsst_y10']['name'] += " (Free Scat.)"
custom_free_scenarios['lsst_y10_om_prior']['name'] += " (Free Scat.)"

fig5 = plot_dspl_corner(
    scenarios=free_scenarios, 
    truth=custom_truth_free, 
    fixed_params=fixed_params, 
    scenarios_to_plot=scenarios_to_plot_fig5,
    custom_ranges=custom_ranges_free,
    latex_labels=latex_labels,
    figsize=(14, 14),
    show_multiple_titles=True,
    custom_linestyles_dict=custom_linestyles_fig5,
    custom_colors_dict=custom_colors_fig5,
    save_path=f"{FIGURE_DIRECTORY}/free_scatter_lsst_y10_forecast_w0waCDM.pdf",
    titles_fontsize=12
)
plt.show()
```
